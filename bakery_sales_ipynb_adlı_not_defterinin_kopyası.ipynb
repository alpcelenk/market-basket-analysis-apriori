```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('bakery_sales_revised.csv')
dataset = dataset[['Transaction', 'Item']]
transactions = dataset.groupby('Transaction')['Item'].apply(list).values.tolist()
```

```python
total_transactions = dataset['Transaction'].nunique()
```

```python
print(total_transactions)
```

```python
from apyori import apriori
rules = apriori(transactions = transactions, min_support = 0.001, min_confidence = 0.18, min_lift = 1.8, min_length = 2, max_length = 2)
```

```python
results = list(rules)
results
```

```python
def inspect(results):
    lhs = []
    rhs = []
    supports = []
    confidences = []
    lifts = []

    for result in results:
        lhs.append(tuple(result[2][0][0])[0])
        rhs.append(tuple(result[2][0][1])[0])
        supports.append(result[1])
        confidences.append(result[2][0][2])
        lifts.append(result[2][0][3])

    return pd.DataFrame({
        'Left Hand Side': lhs,
        'Right Hand Side': rhs,
        'Support': supports,
        'Confidence': confidences,
        'Lift': lifts
    })
```

```python
resultsinDataFrame = inspect(results)

resultsinDataFrame
```
